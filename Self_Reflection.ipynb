## Source

This is a fictional teaching passage, not an external research study.

A university library piloted an AI chatbot for eight weeks to answer questions about opening hours, printing, and room reservations. The chatbot handled 600 questions, and staff reviewed a random sample of 100 responses. Of these, 82 were accurate, 12 were partly accurate, and 6 were incorrect. Students liked the quick responses, but some reported outdated information about holiday hours. The library decided to keep the chatbot for routine questions, update its information weekly, and refer account-specific or complex questions to staff. The pilot did not measure changes in staffing costs or student grades.

## Original user prompt

Summarize the following library chatbot pilot in two sentences for a college student. Use only the provided source. Source: {source}

No separate exercise-specific system prompt was used.

```python
source = 'A university library piloted an AI chatbot for eight weeks to answer questions about opening hours, printing, and room reservations. The chatbot handled 600 questions, and staff reviewed a random sample of 100 responses. Of these, 82 were accurate, 12 were partly accurate, and 6 were incorrect. Students liked the quick responses, but some reported outdated information about holiday hours. The library decided to keep the chatbot for routine questions, update its information weekly, and refer account-specific or complex questions to staff. The pilot did not measure changes in staffing costs or student grades.'
original_prompt = 'Summarize the following library chatbot pilot in two sentences for a college student. Use only the provided source. Source: {source}'.format(source=source)
before = 'The university library tested an AI chatbot for eight weeks to answer routine questions, and students liked its quick responses. The library decided to keep it, update its information weekly, and send more complex questions to staff.'
print("PROMPT:", original_prompt)
print("ORIGINAL SUMMARY:", before)
print("Word count:", len(before.split()))
```

## Critique-and-revise user prompt

Review your original summary against the source below. Check (1) accuracy and unsupported claims, (2) clear language for college students, (3) inclusion of the eight-week pilot, 600 questions, the 100-response audit and its 82/12/6 results, (4) both benefits and limitations, and (5) the library's next steps and the unmeasured outcomes. First provide a brief criteria-based critique. Then revise the summary into exactly one paragraph of 80-100 words. Use only source facts; distinguish the audit sample from all 600 questions. Do not claim cost savings or improved grades. Source: {source}. Original summary: {before}.

```python
reflection_prompt = "Review your original summary against the source below. Check (1) accuracy and unsupported claims, (2) clear language for college students, (3) inclusion of the eight-week pilot, 600 questions, the 100-response audit and its 82/12/6 results, (4) both benefits and limitations, and (5) the library's next steps and the unmeasured outcomes. First provide a brief criteria-based critique. Then revise the summary into exactly one paragraph of 80-100 words. Use only source facts; distinguish the audit sample from all 600 questions. Do not claim cost savings or improved grades. Source: {source}. Original summary: {before}.".format(source=source, before=before)
critique = 'Accuracy: The original summary is supported by the source, but it leaves out important evidence.\nClarity/audience: The wording is accessible to college students.\nEvidence: It omits the 600 questions, 100-response sample, and 82 accurate / 12 partly accurate / 6 incorrect results.\nBalance: It describes fast responses but omits outdated holiday hours.\nNext steps/limits: It should include account-specific referrals and note that costs and grades were not measured.\nLength/format: Expand the 37-word original into one paragraph of 80-100 words.'
after = 'A university library tested an AI chatbot for eight weeks, handling 600 questions about hours, printing, and room reservations. In a random sample of 100 responses, 82 were accurate, 12 were partly accurate, and 6 were incorrect. Students appreciated quick replies, but some reported outdated holiday hours. The library will keep the chatbot for routine questions, update its information weekly, and refer account-specific or complex questions to staff. The pilot did not measure staffing costs or student grades, so it cannot establish improvements in those outcomes.'
print("REFLECTION PROMPT:", reflection_prompt)
print("RECORDED AI CRITIQUE:\n" + critique)
print("IMPROVED SUMMARY:\n" + after)
```

## Check the revision

Mechanical checks cover length, format, and key terms. Accuracy and clarity were also reviewed against the supplied source; these checks alone cannot prove semantic accuracy.

```python
word_count = len(after.split())
assert 80 <= word_count <= 100
assert '\n' not in after
for fact in ['eight weeks', '600', '100', '82', '12', '6', 'holiday', 'weekly', 'account-specific', 'did not measure']:
    assert fact in after, fact
print('Original words:', len(before.split()))
print('Revised words:', word_count)
print('PASS: 80-100 words, one paragraph, and required evidence/limitations present.')
print('Improvement: adds sample-based results, outdated hours, referral scope, and measurement limits.')
```
